# Problema 1

La fábrica de bicicletas Expend produce dos modelos de bicicletas, $M$ y $N$. La matriz $A$ nos indica las cantidades de acero y aluminio que requiere cada bicicleta, expresadas en kg según el modelo a producir, mientras que la matriz $B$ nos entrega un reporte de la cantidad de bicicletas producidas de ambos modelos para los dos primeros meses del año pasado.

<br>

\begin{equation*}
\begin{matrix}
& \begin{matrix}
 & \ \ \ \  M & N
\end{matrix}  \\
&A= \begin{pmatrix}
1 & 2 \\
3 & 2 \\
\end{pmatrix} & \ \begin{matrix}
Acero\\
Aluminio
\end{matrix}\\
\end{matrix}
\ \ \ \ \ \ \ \ \ \ \ \ \ \
\begin{matrix}
& \begin{matrix}
& \ \ \ \ \ \ \ ene & feb &
\end{matrix} & \\
&B= \begin{pmatrix}
290 &  312 \\
 345 &  413 \\
\end{pmatrix} & \begin{matrix}
M\\
N
\end{matrix}\\
\end{matrix}
\end{equation*}

<br>

1. Indique el orden de cada matriz. ¿Es posible determinar la multiplicación entre ambas matrices? Justifique.
2. Determine utilizando *Python* la matriz $C = A \cdot B$
3. Indique e interprete el valor de los elementos $c_{21}$ y $c_{11}$.

```python
#P1:
#A2x2 fila: materiales (ac/al); columnas: modelos (m,n)
#B2x2 fila: modelos (m,n); columnas: meses

#A2x2 * B2x2 = C2x2
#C2x2 fila: materiales (ac/al); columnas: meses (Ene/Feb) **Para ambos modelos**
#Esta matriz proporciona todo el material utilizado en algun mes, para ambos modelos
```

```python
#P2:
import numpy as np
A = np.array([[1,2],[3,2]])
B = np.array([[290,312],[345,413]])

C = A@B #Operador que multiplica matrices es "@" // C = np.dot(A,B).
print(f"C:\n{C}")
```

```python
#P3:
c_11 = 980 #fila 1: acero; columna 1: Enero
print(f"En el mes de enero se utilizo {C[0,0]}kg de acero en la fabricacion de ambos modelos")

c_21 = 1560 #fila 2: aluminio; columna 1: Enero
print(f"En el mes de enero se utilizo {C[1,0]}kg de aluminio en la fabricacion de ambos modelos")
```

# Problema 3

Una fábrica, para un mismo tipo de artículo eléctrico produce $2$ modelos, $A$ y $B$. Las filas de la matriz $M$, en el orden mencionado, nos muestran la cantidad de transistores y resistores que se requieren por cada artículo, según cada modelo. Por otra parte, las filas de la matriz $Q$ nos reportan las cantidades producidas de cada modelo para las tres primeras semanas del mes pasado.

<br>

\begin{split}
M = \begin{pmatrix}
 9 &  14 \\
 4 &  5  \\
\end{pmatrix} \ \ \ \ \ \ \ \ Q = \begin{pmatrix}
 200 &  240 & 220 \\
 175 &  210 & 215 \\
\end{pmatrix}
\end{split}

<br>

Además, considere las matrices $G = M \cdot Q$ y $H = Q \cdot M$.

1. Indique el orden de $M$ y $Q$.
2. ¿Es posible calcular las matrices $G$ y $H$? Justifique.
3. Determine utilizando *Python* la matriz $G$ e interprete los elementos $g_{23}$  y $g_{12}$.
4. ¿Cuántos resistores se utilizaron durante las tres semanas?

```python
#P1:

#M2x2: filas: (elementos(transistores y resistores)); columnas: modelos A y B
#Q2x3: filas: modelos; columnas: semanas (1, 2, 3)


#P2:
#G = M2x2 * Q2x3; Q2x3 = 2 de las filas M y 3 de las columnas de Q; fila: elementos / columnas: semanas (para ambos modelos)
#H = Q2x3 * M2x2; H no se puede efectuar, porque la cantidad de columnas de Q son distintas a la cantidad de filas de M


#P3:
import numpy as np
M = np.array([[9,14],[4,5]])
Q = np.array([[200,240,220],[175,210,215]])

G = M@Q
print(f"G:\n{G}")

g_23 = 1955
print(f"\nLa cantidad de resistores usados durante la 3ra semana para la produccion de ambos modelos fue de {G[1,2]} unidades")
g_12 = 5100
print(f"Se usaron un total de {G[0,1]} transistores para la produccion de de ambos modelos durante la 2da semana")


#P4:
print(f"\nDurante las tres semanas se usaron un total de {sum(G[1,:])} resistores")
```

# Problema 7

Una empresa dedicada al desarrollo de software, que cuenta con tres sucuersales, necesita renovar los siguientes implementos para su personal: laptop de alto rendimiento, monitor externo de alta resolución, teclado mecánico, mouse ergonómico y disco duro de alta capacidad. Para esto, decide cotizar con tres proveedores y así encontrar el precio más conveniente para realizar la compra. La matriz $F$, en el orden mencionado, indica la cantidad de artículos que necesita comprar para cada sucursal. La matriz $G$ indica el precio unitario (en miles de pesos) de cada producto por proveedor.

<br>

\begin{split}
F = \begin{pmatrix}
 7 &  6  &  15 & 17 & 18 \\
 5 &  10  &  20 & 13 & 12 \\
 8 &  4  &  12 & 18 & 9 \\
\end{pmatrix} \ \ \ \ \ \ G = \begin{pmatrix}
 1700 &  1750 &  1800 \\
  868 &  803  &  840  \\
  165 &  160  &  150  \\
  70  &  73  &   65   \\
  110 &  124  &  115  \\
\end{pmatrix}
\end{split}

<br>

1. Calcula utilizando *Python* la matriz $D = F \cdot G$.
2. Interpreta los elementos $d_{33}$ y $d_{21}$.
3. ¿Cuál es el proveedor más conveniente para realizar todas las compras?
4. Si escoge el proveedor más conveniente, ¿en qué sucursal deberá invertir más dinero por la renovación de los implementos?

```python
#F 3x5; (Sucursales) - (Productos)
#G 5x3; (Productos) - (Proveedores)
#D 3x3; (Sucursales) - (Proveedores)

#P1:
import numpy as np
F = np.array([[7,6,15,17,18],[5,10,20,13,12],[8,4,12,18,9]])
G = np.array([[1700,1750,1800],[868,803,840],[165,160,150],[70,73,65],[110,124,115]])

D = F@G
print(f"D:\n{D}")


#P2:
d_33 = 21765
print(f"\nEl costo total de todos los productos de la suscursal 3 en el proveedor 3 es de {D[2,2]} millones de pesos")
d_21 = 22710
print(f"El costo total de todos los productos de la sucursal 2 en el proveedor 1 es de {D[1,0]} millones de pesos")


#P3:
P1 = sum(D[:,0])
P2 = sum(D[:,1])
P3 = sum(D[:,2])
print(f"\n{P1} - {P2} - {P3}")
print("El proveedor 1 es el mas conveniente para realizar todos las comprar, ya que sale mas barato.")



#P4:
print("En la sucursal 1 se debera invertir mas dinero para renovar sus productos")
```
