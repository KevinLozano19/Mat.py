```python
#9:
import numpy as np
import sympy as sp

e = np.e
def G(t):
    return 120- 120 * e **(-0.02 * t)
t = sp.Symbol("t")
ecuacion = sp.Eq(G(t), 75.85)
solucion = sp.solve(ecuacion, t)
print(solucion[0]/12)
```

```python
#10:

import matplotlib.pyplot as plt
def I(t):
    return -0.2 * t + 3
def L(t):
    return -2/15 * t + 8/3

t = np.arange(0,20,0.1)
plt.plot(t, I(t), label ="I(t)")
plt.plot(t, L(t), label ="L(t)")
plt.title('Tasa de interes en duncion del tiempo')
plt.ylabel('Tasa de interes (%)')
plt.xlabel("Tiempo en años")
plt.legend()
plt.grid()
#agregar los ejes al gráfico (OPCIONAL)
ax=plt.gca()
ax.spines[["left","bottom"]].set_position(("data",0))
ax.spines[["right","top"]].set_visible(False)

plt.show()
```

```python
t = sp.Symbol("t")
ecuacion = sp.Eq(L(t), I(t)) #ecuacion
solucion = sp.solve(ecuacion, t)
print(solucion)
```
