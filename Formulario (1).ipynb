```python
#Pregunta 6:
import sympy as sp #Sympy es una libreria que nos ayudara a resolver ecuaciones
x = sp.Symbol("x")
def L(x):
    return 1.85 * x
ecuacion = sp.Eq(L(x), 3480) # Es igual que: (1.85x = 3480)
solucion = sp.solve(ecuacion, x) #Resolver la ecuacion contenida en la variable "ecuacion" con respecto a X
print(f"El tiempo es {solucion[0]:.2f} horas aprox") #Accede al primer elemento de la lista y redondear a 2 decimales
# print(round(solucion[0],2)) Redondear a 2 decimales
```

```python
#1:
#VD: consumo de energia (kWh)
#VI: tiempo (en horas)

#2: numpy
import numpy as np
def E(t):
    return 50 * np.log10(t + 1) + 200 #log base 10 con numpy
print(f"El consumo de energia del Data Center despues de 5 hrs es: {E(5):.1f}kWh")

#2: sympy
import sympy as sp
def E1(t):
    return 50 * sp.log(t + 1, 10) + 200  #Log base 10 con sympy
print(f"El consumo de energia del Data Center despues de 5 hrs es: {E1(5):.1f}kWh")


#3: #Cuanto nos preguntan por la entrada y nos dan un resultado, hacemos una ecuacion
t=sp.Symbol("t")
ecuacion = sp.Eq(E1(t),350) 
solucion = sp.solve(ecuacion, t)
print(f"Deben transcurrir {solucion[0]}hrs, para un consumo de 350kWh")


#4: Para graficar usamos la funcion de numpy
import matplotlib.pyplot as plt
def E(t):
    return 50 * np.log10(t + 1) + 200
t=np.arange(0, 1000, 0.1)
plt.plot(t,E(t), label = "E(t)")
plt.title("Consumo de energia en funcion del tiempo") #VD en funcion de VI
plt.ylabel("Consumo de energia (kWh)")
plt.xlabel("Tiempo (hrs)")
plt.legend()
plt.grid()
#Agregar los ejes al grafico
ax=plt.gca()
ax.spines[["left","bottom"]].set_position(("data",0))
ax.spines[["right","top"]].set_visible(False)
plt.ylim(0, 350)
plt.show()
```

Funcion Lineal


Forma Algebraica de la funcion: f(x) = mx + n


La informacion que entrega m:
los valores de a funcion cambiaran en m unidades por cada unidad que aumente el valor en x. Ejemplo: f(x) = 5x+3: m: indica que por cada unidad que aumente x, la funcion aumentara en 5.


La pendiente "m": nos indica el grado de inclinacion de la recta.
si m > 0 la recta sera CRECIENTE.
si m < 0 la recta sera DECRECIENTE.


El coeficiente de posicion "n": nos indica el corte con el eje Y (vertical), donde las coordenadas de ese punto, seran, (0, n).
