# Problema 7

Por lo general, los proyectos son organizados en varias etapas que deben completarse secuencialmente.

La fase inicial de un proyecto de software tiene una duración de $3$ meses y cada fase subsecuente reduce su duración en un $10 \%$ debido a la experiencia ganada. Determine:

1. ¿Cuánto dura la quinta fase?
2. Calcula la duración total del proyecto después de $10$ fases.

```python
#Sucesion geometrica
#P1:
a1 = 3
R = 0.9 #Ya que la fase siguiente es 10% menor

a5 = a1 * R ** (5-1)
print(f"La duracion de la quinta fase es de {a5:.2f} meses")


#P2:
def a(n):
    return a1 * R ** (n - 1)
lista = []
for i in range(1, 11):
    lista.append(a(i))
print(f"La duracion total del proyecto luego de 10 fases es de {sum(lista):.2f} meses")
```

# Problema 9

El tercer término de una sucesión geométrica es $5$ y el sexto término es $40$. Determine:

1. La razón geométrica de la sucesión.
2. El primer término de la sucesión.
3. La expresión del término de lugar $n$.
4. Utilice un código en *Python* para calcular el término que ocupa el lugar $8$.
4. ¿Qué lugar ocupa en esta sucesión el término $20.480$? Calcule utilizando un ciclo *while*.

```python
a3 = 5
a6 = 40

#P1:
#Razon
R = (a6 / a3) ** (1/3) #Raiz de 3
print(f"La razon es {R}")


#P2: calcular termino de suceccion
#Calcular con el termino mayor
a1 = a6 * R ** (1-6) #-5
print(f"El primer termino es {a1}")

#Calcular con el termino menor
a1 = a3 * R ** (1-3) #-3
print(f"El primer termino es {a1}")


#P3: 
#La expresion del termino de lugar n es a_n = a_1 * R ** (n - 1)
print("La expresion del termino n es: a_n = 1.25 * 2 ** (n - 1)")


#P4:
#Forma 1: sin funcion
a8 = a6 * R ** (8-6)
print(f"El termino de posicion 8 es: {a8}")

#Forma 2: con la funcion
def a(n):
    return 1.25 * 2 ** (n-1)
print(f"El termino de posicion 8 es: {a(8)}")


#P5:
contador = 0
while a(contador) < 20460:
    contador += 1
print(f"En 20480 ocupado la posicion {contador}")

#Para validar que sea correcto
print(f"{a(contador):.0f}")
```

# Problema 10

Una empresa de tecnología está optimizando sus servidores. Para ello mide las transacciones por segundo (TPS) de sus servidores en cada mes. Se ha obtenido que en el segundo mes, el rendimiento del servidor fue $1,2$ miles de TPS y en el quinto mes, el rendimiento fue $2,7$ miles de TPS. Si el número de TPS que hay mes a mes sigue una sucesión geométrica, determina utilizando un código en *Python*:

1. ¿Cuál es el rendimiento el noveno mes?
2. ¿En qué mes el rendimiento es de $23,4$ miles de TPS?

Utilice dos decimales para el cálculo.

```python
#Sucesion geometrica
a2 = 1.2
a5 = 2.7

R = (a5 / a2) ** (1/3)
print(f"Razon: {R:.2f}")

#P1:
a9 = a5 * R ** (9-5)
print(f"El rendimiento el noveno mes sera de: {a9:.2f} miles de TPS")



#P2:
a1 = a2 * R ** (1-2)

def a(n):
    return a1 * R ** (n-1)

contador = 0
while a(contador) < 23.4:
    contador += 1
print(f"El rendimiento sera de 23.4 miles de TPS en el mes {contador}")

#Validar
print(a(13))

```

# Problema 11

Una aplicación de redes sociales se estrenó el $1$ de enero con una cierta cantidad de usuarios. A partir de ahí, el número de usuarios nuevos de cada día fue siempre el mismo. El $3$ de enero tenía $500$ usuarios en total y el $10$ de enero tenía $1.501$ usuarios totales.Determine:

1. ¿Con cuántos usuarios comenzó a funcionar la aplicación?
2. ¿Cuántos usuarios tendrá la aplicación el $31$ de enero?
3. Si la aplicación tiene $5.505$ usuarios, ¿cuántos meses lleva funcionando?



Corresponde a una sucesion aritmetica porque existe una diferencia constante entre los terminos consecutivos.

Recordar la formula: a_n = a_1+(n-1)*d

Pero si tenemos dos terminos no consecutivos calculamos la diferencia de la siguiente forma:
Ejemplo a_10 y a_3 conocidos, entonces, d = (a_10-a_3)/(10-3)

En terminos generales: si a_n y a_k son conocidos, con n > k, la diferencia se calcula como:
d = (a_n - a_k)/(n - k)

```python
#P1:
a_3 = 500
a_10 = 1501

d = (a_10 - a_3) / (10 - 3)

a_1 = a_3 + (-2) * d

print(f"La aplicacion comenzo a funcionar en enero con {a_1} usuarios.")
```

```python
#P2:
a_31 = a_10 + (21) * d
print(a_31)

a_31 = a_3 + (28) * d
print(a_31)

#Solucion creando la funcion
def a(n):
    return a_1 + (n - 1) * d
print(f"La aplicacion el 31 de enero tendra {a(31)} usuarios")
```

```python
#P3:
#a_n = 5505
contador = 0
while a(contador) < 5505:
    contador += 1
print(f"Deben transcurrir {contador} dias")
```

# Problema 12

Una empresa de publicidad ofrece para un puesto de trabajo un salario inicial y un aumento mensual en una razón constante. Javier se presenta a la entrevista, y el encargado del departamento de recursos humanos le explica que en el segundo mes de trabajo su sueldo será de $\$402.000$, y que en el tercer mes de trabajo será de $\$404.010$. Determine:

1. El salario aproximado que Javier recibiría en el décimo mes de trabajo en la empresa.
2. ¿Cuál sería el ingreso total recibido por los sueldos de Javier durante el tercer año de trabajo?

Corresponde a una S.G o Sucesion Geometrica porque hay un aumento en una razón constante

a_2 = 402000
a_3 = 404010

#R=¿? --> a_3 / a_2 "Termino de posicion mayor dividido entre el menor"
R = a_3 / a_2

```python
#P1:
a_2 = 402000
a_3 = 404010

R = a_3 / a_2

a_10 = a_2 * R **(8) #8 = 10 - 2
a_10 = a_3 * R **(10 - 3)
print(f"El 10mo mes recibira ${a_10:.0f} pesos")
```

```python
#P2:
a_1 = a_2 * R ** (-1)
def a(n):
    return a_1 * R ** (n-1)
lista = []
for i in range(25, 37):
    lista.append(a(i))
print(f"Durante el tercer año recibira un total de ${sum(lista):.0f} pesos, aprox")
```

# Problema 13

Una empresa de ciberseguridad capacita trabajadores cada trimestre. En el primer trimestre capacitaron a $30$ trabajadores, en el segundo trimestre a $60$ trabajadores, y en el tercer trimestre a $120$ trabajadores. Si en los siguientes trimestres la empresa mantiene el mismo patrón de crecimiento, determine utilizando *Python*:

1. ¿A cuántos trabajadores capacitarán en el octavo trimestre?
2. Si la empresa lleva en el mercado $3$ años, ¿a cuántos trabajadores han capacitado en total?

Sucesion Geometrica: se caracteriza por presentar una razon (division) constante entre dos terminos consecutivos.

Ejemplo: 5, 15, 45, 135, 405

R = 15/5 = 45/15 = 135/45 = 405/135 = 3, R = 3 es la razon constante.

Formula general: $a_n = a_1 * R**(n-1)$

Con el termino n en funcion del termino k: $a_n = a_k * R**(n-k)$

```python
#P1:
a_1 = 30
a_2 = 60
a_3 = 120
R = 2

a_8 = a_1 * R **(8-1)
print(a_8)

a_8 = a_2 * R **(8-2)
print(a_8)

a_8 = a_3 * R **(8-3)
print(a_8)

#Solucion con funcion
def a(n):
    return a_1 * R**(n-1)
print(f"En el octavo trimestre seran {a(8)} capacitaciones.")
```

```python
#P2:
#En 3 años hay 12 trimestres
lista = []
for i in range(1, 13):
    lista.append(a(i))
print(f"En 3 años hay un total de {sum(lista)} capacitaciones")
```

# Problema 14

Al llenar un tambor con agua, se registra que durante el minuto $4$ el tambor recibe $128$ litros, y durante el minuto $9$ recibe $88$ litros. Sabiendo que en cada minuto recibe una cantidad $d$ de litros de agua menos que en el minuto anterior, determine:

1. ¿En qué instante el tambor deja de recibir agua?
2. ¿Cuántos litros recibe en total el tambor?

Realice los cálculos anteriores utilizando *Python*.

```python
#Sucesion aritmetica
a_4 = 128
a_9 = 88

#P1:
d = (a_9 - a_4) / (9-4) #Diferencia
print(f"diferencia: {d}")

a_1 = a_4 + (1-4) * d

def a(n):
    return a_1 + (n-1) * d

contador = 0
while a(contador) > 0:
    contador += 1
print(f"El tambor deja de recibir agua en el minuto {contador}")


#P2:
lista = []
for i in range(1,21):
    lista.append(a(i))

print(f"El tambor tiene en total {sum(lista)} litros")
```
