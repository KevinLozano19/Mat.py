\begin{split}

4. Un trabajador indica que, en la sucursal japonesa, el costo de producción del producto $3$, modelo $4$ de computador, es más alto que en la sucursal china. ¿Tiene razón? Justifique apropiadamente.

```python
#P1:
print("El orden de J y C es: 5 x 3")


#P2:
#J_21: 120 | J -> Japon; fila2: modelo 2; Columna1: Componente 1.
print("En la sucursal de Japon, los costos de produccion mensual del componente 1 para el modelo 2 es de 120 mil dolares.")
#C_12: 190 | C -> China; fila1: modelo 1; columna2: Componente 2
print("En la sucursal de China, los costos de produccion mensual del componente 2 del modelo 1 es de 190 mil dolares")


#P3: 
import numpy as np
J = np.array([[100,200,150],[120,210,170],[112,200,131],[124,310,170],[136,156,140]])
C = np.array([[110,190,160],[130,220,180],[146,190,141],[143,312,163],[137,116,156]])

T = J + C
print("La matriz del costo total es de: ")
print(T)


#p4:
#J_43: 170 y C_43: 163

j_43 = J[3,2] # -> 4,3 menos 1 por index 0 de Python
c_43 = C[3,2]
print(f"En la sucursal de Japon el costo es {j_43} y en la de China es {c_43} $")
print("Por lo tanto, la afirmacion del trabajador es correcta.")
```
