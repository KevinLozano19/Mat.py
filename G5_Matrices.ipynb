\begin{split}
V = \begin{pmatrix}

# Problema 5

Genere una matriz $M$ de orden $8 \times 9$ donde cada término de la matriz está definido por:

$$m_{ij} = 8i + 12j$$

y calcule su transpuesta.

```python
import numpy as np
M = np.fromfunction(lambda i,j: 8 * (i + 1) + 12 * (j + 1), (8,9))
print(M)
```

MATRIZ TRANSPUESTA: Se consigue intercambiando las filas por las columnas. Por ejemplo, si una matriz es de orden 3x2, su transpuesta sera de orden 2x3

```python
M_T = M.T #Esta es la transpuesta de M
print(M_T)
print(f"\nEl orde de M es: {M.shape}")
print(f"El orde de M_t es: {M_T.shape}")
```

# Problema 6

Genere dos matrices:

*   Una matriz $A$ de orden $5 \times 7$, con $a_{ij} = i \cdot j.$
*   Una matriz $B$ de orden $7 \times 5$, con $b_{ij} = 15i - 2j.$

y calcule $C = A^{T} + \dfrac{1}{2}B$.



```python
import numpy as np
A = np.fromfunction(lambda i,j: (i + 1) * (j + 1), (5,7))
B = np.fromfunction(lambda i,j: 15 * (i + 1) - 2 * (j + 1), (7,5))
print(A)
print(B)

A_T = A.T
C = A_T + 1/2 * B
print(f"\nLa matriz C es: \n{C}")
```
